**Лекция: Условный оператор IF в Python**

---

### Введение

Условный оператор – это ключевая концепция в программировании. Он позволяет программе принимать решения на основе определенных условий и выполнять различные действия в зависимости от того, выполнено условие или нет.

---

### Основная структура `if`

Базовая структура `if` в Python выглядит следующим образом:

```python
if условие:
    # код, который выполнится, если условие истинно
```

Пример:

```python
x = 5

if x > 3:
    print("x больше 3")
```

В этом примере, так как x действительно больше 3, на экран будет выведено "x больше 3".

---

### `else`

Вы можете добавить блок `else` после `if`, чтобы определить действия, которые следует выполнить, если условие в `if` не истинно.

Пример:

```python
x = 2

if x > 3:
    print("x больше 3")
else:
    print("x не больше 3")
```

---

### `elif`

Если у вас есть несколько условий, которые нужно проверить, вы можете использовать `elif` (сокращение от "else if").

Пример:

```python
x = 5

if x > 10:
    print("x больше 10")
elif x > 5:
    print("x больше 5, но не больше 10")
else:
    print("x не больше 5")
```

---

### Вложенные условные операторы

Условный оператор может содержать другие условные операторы. Это называется вложением.

Пример:

```python
x = 5
y = 10

if x == 5:
    if y == 10:
        print("x равно 5 и y равно 10")
```

---

### Краткая запись

Если ваш `if` состоит из одной короткой строки кода, вы можете записать его в одну строку.

Пример:

```python
x = 5

if x > 3: print("x больше 3")
```

---

### Логические операторы

Чтобы проверить несколько условий одновременно, вы можете использовать логические операторы `and`, `or` и `not`.

Пример:

```python
x = 5

if x > 3 and x < 10:
    print("x больше 3 и меньше 10")
```
