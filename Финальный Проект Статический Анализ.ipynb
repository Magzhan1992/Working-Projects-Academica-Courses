**Задача12**<br>
Проведите АБ тест с уровнем значимости 5% что средняя выручка у продукта1 и у продукта2 равны

```python
#A/B-тест: равны ли средние выручки двух продуктов?
#Проверяем гипотезы:
- \(H_0: \mu_1 = \mu_2\)
- \(H_1: \mu_1 \ne \mu_2\)#
```

```python
alpha = 0.05
```

```python
# Двухвыборочный t-тест Уэлча
t_stat, p_value = ttest_ind(
    product1,
    product2,
    equal_var=False
)
```

```python
mean1 = np.mean(product1)
mean2 = np.mean(product2)

print("Средняя выручка Product 1:", mean1)
print("Средняя выручка Product 2:", mean2)

print("Разница средних:", mean1 - mean2)
print("t-статистика:", t_stat)
print("p-value:", p_value)
```

```python
if p_value < alpha:
    print(
        "Отвергаем H0: средние выручки "
        "статистически значимо различаются."
    )
else:
    print(
        "Не отвергаем H0: статистически значимого "
        "различия средних не обнаружено."
    )
```

```python
#Используется двухвыборочный t-тест Уэлча, который не требует равенства дисперсий. Он подходит для независимых выборок. 
#Если оба продукта наблюдались в одни и те же даты и дневные значения связаны общими факторами, следует дополнительно рассмотреть парный t-тест.
```
