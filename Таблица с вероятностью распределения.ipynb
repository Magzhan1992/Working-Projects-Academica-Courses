```python
#Задание 1 - Исследование нормальных распределений#
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Для воспроизводимости результатов#
```

```python
np.random.seed(42)
```

```python
# 1. Генерация двух выборок по 1000 случайных чисел#
```

```python
data1 = np.random.normal(loc=0, scale=1, size=1000)
data2 = np.random.normal(loc=8, scale=2, size=1000)
```

```python
# 2. Построение нормализованных гистограмм#
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(data1, bins=30, density=True, alpha=0.7,
         color='blue', edgecolor='black')
plt.title('Стандартное нормальное распределение')
plt.xlabel('Значение')
plt.ylabel('Плотность вероятности')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(data2, bins=30, density=True, alpha=0.7,
         color='orange', edgecolor='black')
plt.title('Нормальное распределение (μ=8, σ=2)')
plt.xlabel('Значение')
plt.ylabel('Плотность вероятности')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 3. Стандартизация второго распределения#
```

```python
data2_standardized = (data2 - np.mean(data2)) / np.std(data2)
```

```python
# 4. Сравнение стандартизированных данных с первым распределением#
```

```python
plt.figure(figsize=(10, 6))

plt.hist(data1, bins=30, density=True, alpha=0.6,
         color='blue', edgecolor='black',
         label='Стандартное распределение')

plt.hist(data2_standardized, bins=30, density=True, alpha=0.6,
         color='orange', edgecolor='black',
         label='Стандартизированные данные')

plt.title('Сравнение нормальных распределений')
plt.xlabel('Значение')
plt.ylabel('Плотность вероятности')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# 5. Проверка среднего и стандартного отклонения#
```

```python
print('Исходное распределение:')
print('Среднее:', np.mean(data1))
print('Стандартное отклонение:', np.std(data1))

print('\nСтандартизированное распределение:')
print('Среднее:', np.mean(data2_standardized))
print('Стандартное отклонение:', np.std(data2_standardized))
```

```python
#Задание 2. Распределение роста учащихся и стандартизация#
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Для воспроизводимости результатов
np.random.seed(42)
```

```python
# 1. Генерация выборки роста 1000 учащихся#
```

```python
height = np.random.normal(loc=170, scale=10, size=1000)
```

```python
# 2. Стандартизация данных#
```

```python
height_standardized = (
    (height - np.mean(height)) / np.std(height)
)
```

```python
# 3. Построение гистограмм#
plt.figure(figsize=(12, 5))
```

```python
# Гистограмма исходных данных#
```

```python
plt.subplot(1, 2, 1)
plt.hist(height, bins=30, density=True, alpha=0.7,
         color='green', edgecolor='black')
plt.title('Распределение роста учащихся')
plt.xlabel('Рост (см)')
plt.ylabel('Плотность вероятности')
plt.grid(alpha=0.3)
```

```python
# Гистограмма стандартизированных данных#
```

```python
plt.subplot(1, 2, 2)
plt.hist(height_standardized, bins=30, density=True,
         alpha=0.7, color='purple', edgecolor='black')
plt.title('Стандартизированное распределение роста')
plt.xlabel('Стандартизированный рост (Z-score)')
plt.ylabel('Плотность вероятности')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 4. Вывод статистических характеристик#
```

```python
print('Исходные данные:')
print(f'Средний рост: {np.mean(height):.2f} см')
print(f'Стандартное отклонение: {np.std(height):.2f} см')

print('\nСтандартизированные данные:')
print(f'Среднее значение: {np.mean(height_standardized):.6f}')
print(f'Стандартное отклонение: {np.std(height_standardized):.6f}')
```

```python
#Задание 3. Анализ температур и выявление аномалий#
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Для воспроизводимости результатов
np.random.seed(42)

# 1. Генерация 1000 случайных значений температуры
temperature = np.random.normal(
    loc=22,
    scale=3,
    size=1000
)
```

```python
# 2. Определение аномально высоких и низких температур
high_anomalies = temperature > 27
low_anomalies = temperature < 15
```

```python
# 3. Подсчёт количества аномалий
high_count = np.sum(high_anomalies)
low_count = np.sum(low_anomalies)
```

```python
# 4. Расчёт процентов
high_percent = high_count / len(temperature) * 100
low_percent = low_count / len(temperature) * 100

normal_count = len(temperature) - high_count - low_count
normal_percent = normal_count / len(temperature) * 100
```

```python
# 5. Вывод результатов
print('Анализ среднесуточных температур за июль')
print(f'Всего наблюдений: {len(temperature)}')

print(f'\nАномально высокие температуры (> 27°C):')
print(f'Количество дней: {high_count}')
print(f'Процент: {high_percent:.2f}%')

print(f'\nАномально низкие температуры (< 15°C):')
print(f'Количество дней: {low_count}')
print(f'Процент: {low_percent:.2f}%')

print(f'\nТемпературы в пределах заданных порогов:')
print(f'Количество наблюдений: {normal_count}')
print(f'Процент: {normal_percent:.2f}%')
```

```python
# 6. Построение гистограммы с выделением аномалий
plt.figure(figsize=(12, 6))
```

```python
# Нормальные температуры
plt.hist(
    temperature[~high_anomalies & ~low_anomalies],
    bins=30,
    alpha=0.7,
    color='green',
    edgecolor='black',
    label='Нормальные температуры'
)
```

```python
# Аномально высокие температуры
plt.hist(
    temperature[high_anomalies],
    bins=30,
    alpha=0.8,
    color='red',
    edgecolor='black',
    label='Аномально высокие (> 27°C)'
)
```

```python
# Аномально низкие температуры
plt.hist(
    temperature[low_anomalies],
    bins=30,
    alpha=0.8,
    color='blue',
    edgecolor='black',
    label='Аномально низкие (< 15°C)'
)
```

```python
# Пороговые линии
plt.axvline(
    27,
    color='darkred',
    linestyle='--',
    linewidth=2,
    label='Порог высокой температуры: 27°C'
)
```

```python
plt.axvline(
    15,
    color='darkblue',
    linestyle='--',
    linewidth=2,
    label='Порог низкой температуры: 15°C'
)
```

```python
plt.title('Распределение среднесуточных температур за июль')
plt.xlabel('Температура (°C)')
plt.ylabel('Количество наблюдений')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
